```python
# ==========================================================
# ConvNeXt-Tiny + RL-based LR Adaptation — Kather Dataset
# ONE CELL | KERAS 3 COMPATIBLE | MEMORY OPTIMIZED | WORKING VERSION
# ==========================================================

!pip install -q tensorflow seaborn scikit-learn pillow

import tensorflow as tf
tf.config.run_functions_eagerly(True)

# Set memory growth to prevent OOM
gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError as e:
        print(e)

import os, shutil, random, numpy as np
from google.colab import drive
from PIL import Image

# ----------------------------------------------------------
# Mount Drive
# ----------------------------------------------------------
drive.mount('/content/drive')

ZIP_PATH = "/content/drive/MyDrive/archive.zip"
EXTRACT = "/content/Extracted_Kather"

# ----------------------------------------------------------
# Unzip Dataset
# ----------------------------------------------------------
if os.path.exists(EXTRACT):
    shutil.rmtree(EXTRACT)
os.makedirs(EXTRACT, exist_ok=True)

!unzip -q -o "$ZIP_PATH" -d "$EXTRACT"

# ----------------------------------------------------------
# Detect class root
# ----------------------------------------------------------
SOURCE = None
for r, d, _ in os.walk(EXTRACT):
    if len(d) >= 6:
        img_count = sum(len(f) for _,_,f in os.walk(r))
        if img_count > 2000:
            SOURCE = r
            break

print("✅ Class root:", SOURCE)

if SOURCE is None or not os.path.exists(SOURCE):
    raise ValueError(f"Could not find valid source directory!")

# List classes found
classes_found = sorted([d for d in os.listdir(SOURCE) if os.path.isdir(os.path.join(SOURCE, d))])
print(f"✅ Classes found: {classes_found}")
print(f"✅ Number of classes: {len(classes_found)}")

# ----------------------------------------------------------
# Check file extensions in source
# ----------------------------------------------------------
print("\nChecking file extensions in dataset...")
sample_files = []
for cls in classes_found[:2]:
    cls_path = os.path.join(SOURCE, cls)
    files = os.listdir(cls_path)[:3]
    sample_files.extend([(cls, f) for f in files])

for cls, fname in sample_files:
    ext = os.path.splitext(fname)[1].lower()
    print(f"  {cls}/{fname} → extension: {ext}")

# ----------------------------------------------------------
# Reproducibility
# ----------------------------------------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ----------------------------------------------------------
# Train / Val / Test Split WITH FORMAT CONVERSION
# ----------------------------------------------------------
SPLIT = "/content/convnext_split"
TRAIN, VAL, TEST = [f"{SPLIT}/{x}" for x in ["train","val","test"]]

if os.path.exists(SPLIT):
    shutil.rmtree(SPLIT)

os.makedirs(TRAIN, exist_ok=True)
os.makedirs(VAL, exist_ok=True)
os.makedirs(TEST, exist_ok=True)

SOURCE_EXT = (".jpg", ".png", ".jpeg", ".tif", ".bmp", ".tiff")

def collect_imgs(p):
    """Collect all image files from a directory"""
    imgs = []
    if not os.path.exists(p):
        return imgs

    for r, _, fs in os.walk(p):
        for f in fs:
            if f.lower().endswith(SOURCE_EXT):
                imgs.append(os.path.join(r, f))
    return imgs

def convert_and_copy(src_path, dst_path):
    """Copy image and convert to .jpg if necessary"""
    try:
        _, ext = os.path.splitext(src_path)
        ext = ext.lower()

        if ext in ['.tif', '.tiff']:
            img = Image.open(src_path)
            if img.mode != 'RGB':
                img = img.convert('RGB')
            dst_path = os.path.splitext(dst_path)[0] + '.jpg'
            img.save(dst_path, 'JPEG', quality=95)
        else:
            shutil.copy2(src_path, dst_path)

        return True
    except Exception as e:
        print(f"  Error processing {src_path}: {e}")
        return False

rng = np.random.default_rng(SEED)

print("\n" + "="*60)
print("Splitting dataset into train/val/test...")
print("="*60)

total_train = 0
total_val = 0
total_test = 0
conversion_count = 0

for cls in classes_found:
    class_path = os.path.join(SOURCE, cls)

    if not os.path.isdir(class_path):
        continue

    files = collect_imgs(class_path)

    if len(files) == 0:
        print(f"⚠️  Warning: No images found in class '{cls}'")
        continue

    idx = rng.permutation(len(files))
    n = len(files)

    tr = int(0.65 * n)
    va = int(0.20 * n)

    os.makedirs(f"{TRAIN}/{cls}", exist_ok=True)
    os.makedirs(f"{VAL}/{cls}", exist_ok=True)
    os.makedirs(f"{TEST}/{cls}", exist_ok=True)

    train_count = 0
    val_count = 0
    test_count = 0

    for i, j in enumerate(idx):
        src_file = files[j]
        base_name = os.path.basename(src_file)

        if i < tr:
            dst = f"{TRAIN}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                train_count += 1
        elif i < tr + va:
            dst = f"{VAL}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                val_count += 1
        else:
            dst = f"{TEST}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                test_count += 1

        if src_file.lower().endswith(('.tif', '.tiff')):
            conversion_count += 1

    total_train += train_count
    total_val += val_count
    total_test += test_count

    print(f"  {cls}: {n} total → Train: {train_count}, Val: {val_count}, Test: {test_count}")

print(f"\n✅ Split complete!")
print(f"   Total - Train: {total_train}, Val: {total_val}, Test: {total_test}")
print(f"   Converted {conversion_count} TIF/TIFF files to JPG")

print("\nVerifying splits...")
for split_name, split_path in [("Train", TRAIN), ("Val", VAL), ("Test", TEST)]:
    all_files = []
    for r, _, files in os.walk(split_path):
        all_files.extend(files)

    ext_counts = {}
    for f in all_files:
        ext = os.path.splitext(f)[1].lower()
        ext_counts[ext] = ext_counts.get(ext, 0) + 1

    print(f"  {split_name}: {len(all_files)} images - Extensions: {ext_counts}")

# ----------------------------------------------------------
# Create tf.data.Dataset
# ----------------------------------------------------------
IMG_SIZE = (192, 192)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE

def create_dataset(subset_path, training=False):
    """Create a TensorFlow dataset from a directory"""

    if not os.path.exists(subset_path):
        raise ValueError(f"Directory does not exist: {subset_path}")

    num_images = sum(len(files) for _, _, files in os.walk(subset_path))
    print(f"Creating dataset from {subset_path} ({num_images} images)...")

    base_dataset = tf.keras.utils.image_dataset_from_directory(
        subset_path,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode='categorical',
        shuffle=training,
        seed=SEED,
        interpolation='bilinear'
    )

    # Store class names from base dataset
    class_names = base_dataset.class_names

    # Apply preprocessing for ConvNeXt
    def preprocess(images, labels):
        images = tf.keras.applications.convnext.preprocess_input(images)
        return images, labels

    dataset = base_dataset.map(preprocess, num_parallel_calls=AUTOTUNE)

    if training:
        augmentation = tf.keras.Sequential([
            tf.keras.layers.RandomRotation(0.1),
            tf.keras.layers.RandomZoom(0.1),
            tf.keras.layers.RandomFlip("horizontal"),
        ])

        def augment(images, labels):
            images = augmentation(images, training=True)
            return images, labels

        dataset = dataset.map(augment, num_parallel_calls=AUTOTUNE)

    dataset = dataset.prefetch(AUTOTUNE)

    # Attach class_names to the dataset
    dataset.class_names = class_names

    return dataset

# Create datasets
print("\n" + "="*60)
print("Creating TensorFlow datasets...")
print("="*60)

train_dataset = create_dataset(TRAIN, training=True)
val_dataset = create_dataset(VAL, training=False)
test_dataset = create_dataset(TEST, training=False)

class_names = train_dataset.class_names
num_classes = len(class_names)
print(f"\n✅ Number of classes: {num_classes}")
print(f"✅ Class names: {class_names}")

# ----------------------------------------------------------
# Memory-efficient model definition
# ----------------------------------------------------------
from tensorflow.keras.applications import ConvNeXtTiny
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D, BatchNormalization, Input
from tensorflow.keras import regularizers, optimizers
import tensorflow.keras.backend as K

K.clear_session()

print("\n" + "="*60)
print("Building ConvNeXt-Tiny model...")
print("="*60)

def create_model():
    base_model = ConvNeXtTiny(
        weights="imagenet",
        include_top=False,
        input_shape=IMG_SIZE + (3,),
        pooling=None
    )

    base_model.trainable = False

    inputs = Input(shape=IMG_SIZE + (3,))
    x = base_model(inputs, training=False)
    x = GlobalAveragePooling2D()(x)
    x = BatchNormalization()(x)
    x = Dense(256, activation="relu", kernel_regularizer=regularizers.l2(1e-5))(x)
    x = Dropout(0.4)(x)
    x = Dense(128, activation="relu")(x)
    x = Dropout(0.2)(x)
    outputs = Dense(num_classes, activation="softmax")(x)

    model = Model(inputs=inputs, outputs=outputs)
    return model, base_model

model, base_model = create_model()

initial_lr = 1e-4
model.compile(
    optimizer=optimizers.Adam(learning_rate=initial_lr),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print(f"\nModel created! Total params: {model.count_params():,}")

# ----------------------------------------------------------
# RL-based LR Controller (FIXED for Keras 3)
# ----------------------------------------------------------
class RLLRController:
    def __init__(self, lr=1e-4, clip=0.15, bounds=(1e-6, 1e-3)):
        self.lr = float(lr)
        self.clip = clip
        self.bounds = bounds
        self.prev_reward = 0.0

    def update(self, reward):
        reward = float(reward)
        advantage = reward - self.prev_reward
        adjustment_factor = 1 + np.clip(advantage, -self.clip, self.clip)
        self.lr *= adjustment_factor
        self.lr = float(np.clip(self.lr, self.bounds[0], self.bounds[1]))
        self.prev_reward = reward
        return self.lr

rl_controller = RLLRController(lr=initial_lr)

# ----------------------------------------------------------
# Training Callbacks (FIXED for Keras 3)
# ----------------------------------------------------------
class RLLRCallback(tf.keras.callbacks.Callback):
    def __init__(self, rl_controller):
        super().__init__()
        self.rl_controller = rl_controller

    def on_epoch_end(self, epoch, logs=None):
        if logs is None:
            return

        val_acc = logs.get('val_accuracy', 0)
        val_loss = logs.get('val_loss', 0)
        reward = 0.7 * val_acc + 0.3 * (1 - val_loss)
        new_lr = self.rl_controller.update(reward)

        # FIXED: Keras 3 compatible way to set learning rate
        self.model.optimizer.learning_rate.assign(new_lr)

        print(f"  RL Update: Reward={reward:.4f}, New LR={new_lr:.2e}")

# ----------------------------------------------------------
# Phase 1: Frozen Backbone Training
# ----------------------------------------------------------
print("\n" + "="*60)
print("Phase 1: Training with Frozen Backbone (15 epochs)")
print("="*60)

callbacks = [
    RLLRCallback(rl_controller),
    tf.keras.callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-6,
        verbose=1
    )
]

history1 = model.fit(
    train_dataset,
    epochs=15,
    validation_data=val_dataset,
    callbacks=callbacks,
    verbose=1
)

# ----------------------------------------------------------
# Phase 2: Fine-tuning
# ----------------------------------------------------------
print("\n" + "="*60)
print("Phase 2: Fine-tuning (10 epochs)")
print("="*60)

base_model.trainable = True
for layer in base_model.layers[:-40]:
    layer.trainable = False

print(f"Trainable layers: {sum(1 for l in base_model.layers if l.trainable)}/{len(base_model.layers)}")

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

rl_controller = RLLRController(lr=1e-5)
callbacks[0] = RLLRCallback(rl_controller)

history2 = model.fit(
    train_dataset,
    epochs=10,
    validation_data=val_dataset,
    callbacks=callbacks,
    verbose=1
)

# ----------------------------------------------------------
# Test Evaluation
# ----------------------------------------------------------
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

print("\n" + "="*60)
print("Test Evaluation")
print("="*60)

test_loss, test_acc = model.evaluate(test_dataset, verbose=1)
print(f"\n✅ Test Loss: {test_loss:.4f}")
print(f"✅ Test Accuracy: {test_acc:.4f}")

print("\nGenerating predictions...")
y_pred_probs = model.predict(test_dataset, verbose=1)
y_pred = np.argmax(y_pred_probs, axis=1)

y_true = np.concatenate([y for x, y in test_dataset], axis=0)
y_true = np.argmax(y_true, axis=1)

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names,
            yticklabels=class_names)
plt.title("Confusion Matrix", fontsize=16)
plt.ylabel("True Label", fontsize=12)
plt.xlabel("Predicted Label", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Classification Report
print("\n" + "="*60)
print("Classification Report")
print("="*60)
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))

# Plot training history
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
axes[0].plot(history1.history['accuracy'], label='Train Acc (Phase 1)', alpha=0.7)
axes[0].plot(history1.history['val_accuracy'], label='Val Acc (Phase 1)', alpha=0.7)
if len(history2.history['accuracy']) > 0:
    axes[0].plot(range(len(history1.history['accuracy']),
                     len(history1.history['accuracy']) + len(history2.history['accuracy'])),
                 history2.history['accuracy'], label='Train Acc (Phase 2)', alpha=0.7)
    axes[0].plot(range(len(history1.history['val_accuracy']),
                     len(history1.history['val_accuracy']) + len(history2.history['val_accuracy'])),
                 history2.history['val_accuracy'], label='Val Acc (Phase 2)', alpha=0.7)
axes[0].axvline(x=len(history1.history['accuracy'])-1, color='r', linestyle='--', alpha=0.5, label='Phase Transition')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].set_title('Training and Validation Accuracy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Loss
axes[1].plot(history1.history['loss'], label='Train Loss (Phase 1)', alpha=0.7)
axes[1].plot(history1.history['val_loss'], label='Val Loss (Phase 1)', alpha=0.7)
if len(history2.history['loss']) > 0:
    axes[1].plot(range(len(history1.history['loss']),
                     len(history1.history['loss']) + len(history2.history['loss'])),
                 history2.history['loss'], label='Train Loss (Phase 2)', alpha=0.7)
    axes[1].plot(range(len(history1.history['val_loss']),
                     len(history1.history['val_loss']) + len(history2.history['val_loss'])),
                 history2.history['val_loss'], label='Val Loss (Phase 2)', alpha=0.7)
axes[1].axvline(x=len(history1.history['loss'])-1, color='r', linestyle='--', alpha=0.5, label='Phase Transition')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].set_title('Training and Validation Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# Save Model
# ----------------------------------------------------------
model.save("/content/convnext_RL_kather_model.keras")
print("\n✅ Model saved: convnext_RL_kather_model.keras")

model.save_weights("/content/convnext_RL_kather_weights.h5")
print("✅ Weights saved: convnext_RL_kather_weights.h5")

print("\n" + "="*60)
print("Training Complete!")
print("="*60)
print(f"\nFinal Performance:")
print(f"- Test Accuracy: {test_acc:.4f}")
print(f"- Test Loss: {test_loss:.4f}")
print(f"- Number of classes: {num_classes}")
print(f"- Best val accuracy: {max(history1.history['val_accuracy'] + history2.history['val_accuracy']):.4f}")
```

```python
# ==========================================================
# ConvNeXt-Tiny + RL-based LR Adaptation — Kather Dataset
# ONE CELL | KERAS 3 COMPATIBLE | MEMORY OPTIMIZED | WORKING VERSION
# ==========================================================

!pip install -q tensorflow seaborn scikit-learn pillow

import tensorflow as tf
tf.config.run_functions_eagerly(True)

# Set memory growth to prevent OOM
gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError as e:
        print(e)

import os, shutil, random, numpy as np
from google.colab import drive
from PIL import Image

# ----------------------------------------------------------
# Mount Drive
# ----------------------------------------------------------
drive.mount('/content/drive')

ZIP_PATH = "/content/drive/MyDrive/archive (1).zip"
EXTRACT = "/content/Extracted_Kather"

# ----------------------------------------------------------
# Unzip Dataset
# ----------------------------------------------------------
if os.path.exists(EXTRACT):
    shutil.rmtree(EXTRACT)
os.makedirs(EXTRACT, exist_ok=True)

!unzip -q -o "$ZIP_PATH" -d "$EXTRACT"

# ----------------------------------------------------------
# Detect class root
# ----------------------------------------------------------
SOURCE = None
for r, d, _ in os.walk(EXTRACT):
    if len(d) >= 6:
        img_count = sum(len(f) for _,_,f in os.walk(r))
        if img_count > 2000:
            SOURCE = r
            break

print("✅ Class root:", SOURCE)

if SOURCE is None or not os.path.exists(SOURCE):
    raise ValueError(f"Could not find valid source directory!")

# List classes found
classes_found = sorted([d for d in os.listdir(SOURCE) if os.path.isdir(os.path.join(SOURCE, d))])
print(f"✅ Classes found: {classes_found}")
print(f"✅ Number of classes: {len(classes_found)}")

# ----------------------------------------------------------
# Check file extensions in source
# ----------------------------------------------------------
print("\nChecking file extensions in dataset...")
sample_files = []
for cls in classes_found[:2]:
    cls_path = os.path.join(SOURCE, cls)
    files = os.listdir(cls_path)[:3]
    sample_files.extend([(cls, f) for f in files])

for cls, fname in sample_files:
    ext = os.path.splitext(fname)[1].lower()
    print(f"  {cls}/{fname} → extension: {ext}")

# ----------------------------------------------------------
# Reproducibility
# ----------------------------------------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ----------------------------------------------------------
# Train / Val / Test Split WITH FORMAT CONVERSION
# ----------------------------------------------------------
SPLIT = "/content/convnext_split"
TRAIN, VAL, TEST = [f"{SPLIT}/{x}" for x in ["train","val","test"]]

if os.path.exists(SPLIT):
    shutil.rmtree(SPLIT)

os.makedirs(TRAIN, exist_ok=True)
os.makedirs(VAL, exist_ok=True)
os.makedirs(TEST, exist_ok=True)

SOURCE_EXT = (".jpg", ".png", ".jpeg", ".tif", ".bmp", ".tiff")

def collect_imgs(p):
    """Collect all image files from a directory"""
    imgs = []
    if not os.path.exists(p):
        return imgs

    for r, _, fs in os.walk(p):
        for f in fs:
            if f.lower().endswith(SOURCE_EXT):
                imgs.append(os.path.join(r, f))
    return imgs

def convert_and_copy(src_path, dst_path):
    """Copy image and convert to .jpg if necessary"""
    try:
        _, ext = os.path.splitext(src_path)
        ext = ext.lower()

        if ext in ['.tif', '.tiff']:
            img = Image.open(src_path)
            if img.mode != 'RGB':
                img = img.convert('RGB')
            dst_path = os.path.splitext(dst_path)[0] + '.jpg'
            img.save(dst_path, 'JPEG', quality=95)
        else:
            shutil.copy2(src_path, dst_path)

        return True
    except Exception as e:
        print(f"  Error processing {src_path}: {e}")
        return False

rng = np.random.default_rng(SEED)

print("\n" + "="*60)
print("Splitting dataset into train/val/test...")
print("="*60)

total_train = 0
total_val = 0
total_test = 0
conversion_count = 0

for cls in classes_found:
    class_path = os.path.join(SOURCE, cls)

    if not os.path.isdir(class_path):
        continue

    files = collect_imgs(class_path)

    if len(files) == 0:
        print(f"⚠️  Warning: No images found in class '{cls}'")
        continue

    idx = rng.permutation(len(files))
    n = len(files)

    tr = int(0.65 * n)
    va = int(0.20 * n)

    os.makedirs(f"{TRAIN}/{cls}", exist_ok=True)
    os.makedirs(f"{VAL}/{cls}", exist_ok=True)
    os.makedirs(f"{TEST}/{cls}", exist_ok=True)

    train_count = 0
    val_count = 0
    test_count = 0

    for i, j in enumerate(idx):
        src_file = files[j]
        base_name = os.path.basename(src_file)

        if i < tr:
            dst = f"{TRAIN}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                train_count += 1
        elif i < tr + va:
            dst = f"{VAL}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                val_count += 1
        else:
            dst = f"{TEST}/{cls}/{base_name}"
            if convert_and_copy(src_file, dst):
                test_count += 1

        if src_file.lower().endswith(('.tif', '.tiff')):
            conversion_count += 1

    total_train += train_count
    total_val += val_count
    total_test += test_count

    print(f"  {cls}: {n} total → Train: {train_count}, Val: {val_count}, Test: {test_count}")

print(f"\n✅ Split complete!")
print(f"   Total - Train: {total_train}, Val: {total_val}, Test: {total_test}")
print(f"   Converted {conversion_count} TIF/TIFF files to JPG")

print("\nVerifying splits...")
for split_name, split_path in [("Train", TRAIN), ("Val", VAL), ("Test", TEST)]:
    all_files = []
    for r, _, files in os.walk(split_path):
        all_files.extend(files)

    ext_counts = {}
    for f in all_files:
        ext = os.path.splitext(f)[1].lower()
        ext_counts[ext] = ext_counts.get(ext, 0) + 1

    print(f"  {split_name}: {len(all_files)} images - Extensions: {ext_counts}")

# ----------------------------------------------------------
# Create tf.data.Dataset
# ----------------------------------------------------------
IMG_SIZE = (192, 192)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE

def create_dataset(subset_path, training=False):
    """Create a TensorFlow dataset from a directory"""

    if not os.path.exists(subset_path):
        raise ValueError(f"Directory does not exist: {subset_path}")

    num_images = sum(len(files) for _, _, files in os.walk(subset_path))
    print(f"Creating dataset from {subset_path} ({num_images} images)...")

    base_dataset = tf.keras.utils.image_dataset_from_directory(
        subset_path,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode='categorical',
        shuffle=training,
        seed=SEED,
        interpolation='bilinear'
    )

    # Store class names from base dataset
    class_names = base_dataset.class_names

    # Apply preprocessing for ConvNeXt
    def preprocess(images, labels):
        images = tf.keras.applications.convnext.preprocess_input(images)
        return images, labels

    dataset = base_dataset.map(preprocess, num_parallel_calls=AUTOTUNE)

    if training:
        augmentation = tf.keras.Sequential([
            tf.keras.layers.RandomRotation(0.1),
            tf.keras.layers.RandomZoom(0.1),
            tf.keras.layers.RandomFlip("horizontal"),
        ])

        def augment(images, labels):
            images = augmentation(images, training=True)
            return images, labels

        dataset = dataset.map(augment, num_parallel_calls=AUTOTUNE)

    dataset = dataset.prefetch(AUTOTUNE)

    # Attach class_names to the dataset
    dataset.class_names = class_names

    return dataset

# Create datasets
print("\n" + "="*60)
print("Creating TensorFlow datasets...")
print("="*60)

train_dataset = create_dataset(TRAIN, training=True)
val_dataset = create_dataset(VAL, training=False)
test_dataset = create_dataset(TEST, training=False)

class_names = train_dataset.class_names
num_classes = len(class_names)
print(f"\n✅ Number of classes: {num_classes}")
print(f"✅ Class names: {class_names}")

# ----------------------------------------------------------
# Memory-efficient model definition
# ----------------------------------------------------------
from tensorflow.keras.applications import ConvNeXtTiny
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D, BatchNormalization, Input
from tensorflow.keras import regularizers, optimizers
import tensorflow.keras.backend as K

K.clear_session()

print("\n" + "="*60)
print("Building ConvNeXt-Tiny model...")
print("="*60)

def create_model():
    base_model = ConvNeXtTiny(
        weights="imagenet",
        include_top=False,
        input_shape=IMG_SIZE + (3,),
        pooling=None
    )

    base_model.trainable = False

    inputs = Input(shape=IMG_SIZE + (3,))
    x = base_model(inputs, training=False)
    x = GlobalAveragePooling2D()(x)
    x = BatchNormalization()(x)
    x = Dense(256, activation="relu", kernel_regularizer=regularizers.l2(1e-5))(x)
    x = Dropout(0.4)(x)
    x = Dense(128, activation="relu")(x)
    x = Dropout(0.2)(x)
    outputs = Dense(num_classes, activation="softmax")(x)

    model = Model(inputs=inputs, outputs=outputs)
    return model, base_model

model, base_model = create_model()

initial_lr = 1e-4
model.compile(
    optimizer=optimizers.Adam(learning_rate=initial_lr),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print(f"\nModel created! Total params: {model.count_params():,}")

# ----------------------------------------------------------
# RL-based LR Controller (FIXED for Keras 3)
# ----------------------------------------------------------
class RLLRController:
    def __init__(self, lr=1e-4, clip=0.15, bounds=(1e-6, 1e-3)):
        self.lr = float(lr)
        self.clip = clip
        self.bounds = bounds
        self.prev_reward = 0.0

    def update(self, reward):
        reward = float(reward)
        advantage = reward - self.prev_reward
        adjustment_factor = 1 + np.clip(advantage, -self.clip, self.clip)
        self.lr *= adjustment_factor
        self.lr = float(np.clip(self.lr, self.bounds[0], self.bounds[1]))
        self.prev_reward = reward
        return self.lr

rl_controller = RLLRController(lr=initial_lr)

# ----------------------------------------------------------
# Training Callbacks (FIXED for Keras 3)
# ----------------------------------------------------------
class RLLRCallback(tf.keras.callbacks.Callback):
    def __init__(self, rl_controller):
        super().__init__()
        self.rl_controller = rl_controller

    def on_epoch_end(self, epoch, logs=None):
        if logs is None:
            return

        val_acc = logs.get('val_accuracy', 0)
        val_loss = logs.get('val_loss', 0)
        reward = 0.7 * val_acc + 0.3 * (1 - val_loss)
        new_lr = self.rl_controller.update(reward)

        # FIXED: Keras 3 compatible way to set learning rate
        self.model.optimizer.learning_rate.assign(new_lr)

        print(f"  RL Update: Reward={reward:.4f}, New LR={new_lr:.2e}")

# ----------------------------------------------------------
# Phase 1: Frozen Backbone Training
# ----------------------------------------------------------
print("\n" + "="*60)
print("Phase 1: Training with Frozen Backbone (15 epochs)")
print("="*60)

callbacks = [
    RLLRCallback(rl_controller),
    tf.keras.callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-6,
        verbose=1
    )
]

history1 = model.fit(
    train_dataset,
    epochs=15,
    validation_data=val_dataset,
    callbacks=callbacks,
    verbose=1
)

# ----------------------------------------------------------
# Phase 2: Fine-tuning
# ----------------------------------------------------------
print("\n" + "="*60)
print("Phase 2: Fine-tuning (25 epochs)")
print("="*60)

base_model.trainable = True
for layer in base_model.layers[:-40]:
    layer.trainable = False

print(f"Trainable layers: {sum(1 for l in base_model.layers if l.trainable)}/{len(base_model.layers)}")

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

rl_controller = RLLRController(lr=1e-5)
callbacks[0] = RLLRCallback(rl_controller)

history2 = model.fit(
    train_dataset,
    epochs=25,
    validation_data=val_dataset,
    callbacks=callbacks,
    verbose=1
)

# ----------------------------------------------------------
# Test Evaluation
# ----------------------------------------------------------
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

print("\n" + "="*60)
print("Test Evaluation")
print("="*60)

test_loss, test_acc = model.evaluate(test_dataset, verbose=1)
print(f"\n✅ Test Loss: {test_loss:.4f}")
print(f"✅ Test Accuracy: {test_acc:.4f}")

print("\nGenerating predictions...")
y_pred_probs = model.predict(test_dataset, verbose=1)
y_pred = np.argmax(y_pred_probs, axis=1)

y_true = np.concatenate([y for x, y in test_dataset], axis=0)
y_true = np.argmax(y_true, axis=1)

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names,
            yticklabels=class_names)
plt.title("Confusion Matrix", fontsize=16)
plt.ylabel("True Label", fontsize=12)
plt.xlabel("Predicted Label", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Classification Report
print("\n" + "="*60)
print("Classification Report")
print("="*60)
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))

# Plot training history
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
axes[0].plot(history1.history['accuracy'], label='Train Acc (Phase 1)', alpha=0.7)
axes[0].plot(history1.history['val_accuracy'], label='Val Acc (Phase 1)', alpha=0.7)
if len(history2.history['accuracy']) > 0:
    axes[0].plot(range(len(history1.history['accuracy']),
                     len(history1.history['accuracy']) + len(history2.history['accuracy'])),
                 history2.history['accuracy'], label='Train Acc (Phase 2)', alpha=0.7)
    axes[0].plot(range(len(history1.history['val_accuracy']),
                     len(history1.history['val_accuracy']) + len(history2.history['val_accuracy'])),
                 history2.history['val_accuracy'], label='Val Acc (Phase 2)', alpha=0.7)
axes[0].axvline(x=len(history1.history['accuracy'])-1, color='r', linestyle='--', alpha=0.5, label='Phase Transition')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].set_title('Training and Validation Accuracy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Loss
axes[1].plot(history1.history['loss'], label='Train Loss (Phase 1)', alpha=0.7)
axes[1].plot(history1.history['val_loss'], label='Val Loss (Phase 1)', alpha=0.7)
if len(history2.history['loss']) > 0:
    axes[1].plot(range(len(history1.history['loss']),
                     len(history1.history['loss']) + len(history2.history['loss'])),
                 history2.history['loss'], label='Train Loss (Phase 2)', alpha=0.7)
    axes[1].plot(range(len(history1.history['val_loss']),
                     len(history1.history['val_loss']) + len(history2.history['val_loss'])),
                 history2.history['val_loss'], label='Val Loss (Phase 2)', alpha=0.7)
axes[1].axvline(x=len(history1.history['loss'])-1, color='r', linestyle='--', alpha=0.5, label='Phase Transition')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].set_title('Training and Validation Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# Save Model
# ----------------------------------------------------------
model.save("/content/convnext_RL_kather_model.keras")
print("\n✅ Model saved: convnext_RL_kather_model.keras")

model.save_weights("/content/convnext_RL_kather_weights.h5")
print("✅ Weights saved: convnext_RL_kather_weights.h5")

print("\n" + "="*60)
print("Training Complete!")
print("="*60)
print(f"\nFinal Performance:")
print(f"- Test Accuracy: {test_acc:.4f}")
print(f"- Test Loss: {test_loss:.4f}")
print(f"- Number of classes: {num_classes}")
print(f"- Best val accuracy: {max(history1.history['val_accuracy'] + history2.history['val_accuracy']):.4f}")
```
